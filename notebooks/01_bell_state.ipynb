Basis ordering: `|q1 q0⟩` (Qiskit convention).


```python
from pathlib import Path
import sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
import numpy as np
import matplotlib.pyplot as plt
from src.bell_state import BASIS, statevectors, probabilities, density_matrix, classical_mixture, xx_expectation
```

```python
for stage, ket in statevectors().items():
    print(stage, np.round(ket, 3))
    print(dict(zip(BASIS, np.round(probabilities()[stage], 3))))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 3), sharey=True)
for ax, (stage, probs) in zip(axes, probabilities().items()):
    ax.bar(BASIS, probs, color='#173E32')
    ax.set_title(stage)
    ax.set_ylim(0, 1.05)
    ax.set_ylabel('Probability')
plt.tight_layout()
plt.show()
```

## Same Z-basis probabilities, different X-basis correlations


```python
bell = density_matrix(statevectors()['After CNOT'])
mixture = classical_mixture()
print('Same Z probabilities:', np.allclose(np.diag(bell), np.diag(mixture)))
print('Bell <XX>:', xx_expectation(bell))
print('Classical mixture <XX>:', xx_expectation(mixture))
```
